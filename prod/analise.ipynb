```python
import pandas as pd

df = pd.read_csv("dados_teste/resultados_teste.csv", sep=",")
```

```python
df_filtrado = df[df["acertosCompletos"] > 829]

novoTotal = 898

df_filtrado["acertosCompletosPorcentagem"] = 100 * df_filtrado["acertosCompletos"] / novoTotal
df_filtrado["acertosTop5Porcentagem"] = 100 * df_filtrado["acertosTop5"] / novoTotal
df_filtrado["errosPorcentagem"] = 100 * df_filtrado["erros"] / novoTotal

df_filtrado["tempo_execucao"] = df_filtrado["tempo_execucao"] - ((df_filtrado["tempo_execucao"] / df.iloc[0]["totalTestado"]) * 50)
df_filtrado["totalTestado"] = novoTotal

nova_ordem = ["rerank", "filtroApi", "listaFiltroApis", "modeloChat", "modeloClassificador", "embedding", "acertosCompletos", "acertosCompletosPorcentagem", "acertosTop5", "acertosTop5Porcentagem", "erros", "errosPorcentagem", "totalTestado", "tempo_execucao"]
df_filtrado = df_filtrado[nova_ordem]

df_filtrado = df_filtrado.sort_values("acertosCompletos", ascending=False)
df_filtrado.to_csv("dados_teste/melhores.csv", index=False)
```

```python
colunasInteresses = []
colunaAgrupamento = "rerank"

colunasInteresses.append(colunaAgrupamento)
colunasInteresses.extend(["acertosCompletos", "acertosTop5", "erros", "totalTestado", "tempo_execucao"])

df_agrupado = df[colunasInteresses].groupby(colunaAgrupamento).mean().reset_index()

novoTotal = 898

df_agrupado["acertosCompletosPorcentagem"] = 100 * df_agrupado["acertosCompletos"] / novoTotal
df_agrupado["acertosTop5Porcentagem"] = 100 * df_agrupado["acertosTop5"] / novoTotal
df_agrupado["errosPorcentagem"] = 100 * df_agrupado["erros"] / novoTotal

df_agrupado["tempo_execucao"] = df_agrupado["tempo_execucao"] - ((df_agrupado["tempo_execucao"] / df.iloc[0]["totalTestado"]) * 50)
df_agrupado["totalTestado"] = novoTotal

nova_ordem = [colunaAgrupamento, "acertosCompletos", "acertosCompletosPorcentagem", "acertosTop5", "acertosTop5Porcentagem", "erros", "errosPorcentagem", "totalTestado", "tempo_execucao"]
df_reordenado = df_agrupado[nova_ordem]

df_reordenado.to_csv(f"dados_teste/{colunaAgrupamento}.csv", index=False)
```
